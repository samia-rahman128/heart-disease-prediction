```python
import pandas as pd
df = pd.read_csv("cardio_train.csv", sep=";")
print(df.shape)
df.head()
```

```python
df.info()
df.isnull().sum()
df.duplicated().sum()
```

```python
df.describe()
```

```python
df['cardio'].value_counts()
df['cardio'].value_counts(normalize=True)
```

```python
print((df['ap_hi'] < 0).sum(), (df['ap_hi'] > 300).sum())
print((df['ap_lo'] < 0).sum(), (df['ap_lo'] > 200).sum())
print((df['ap_lo'] > df['ap_hi']).sum())
```

```python
import matplotlib.pyplot as plt
df[['ap_hi','ap_lo','height','weight']].plot(kind='box', subplots=True, layout=(1,4), figsize=(12,4))
plt.show()
```

```python
import seaborn as sns
plt.figure(figsize=(10,8))
sns.heatmap(df.corr(), annot=True, fmt='.2f', cmap='coolwarm')
plt.show()
```
